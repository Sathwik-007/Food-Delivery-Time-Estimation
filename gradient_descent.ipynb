# Learning

```python
import numpy as np
import pandas as pd
```

```python
X_raw = np.array([[500, 1], [1000, 2], [1500, 2], [2000, 3], 
                 [2500, 3], [3000, 4], [4000, 4], [5000, 5]], dtype=float)
y = np.array([150, 200, 250, 300,
              320, 380, 450, 550], dtype=float)
```

```python
X_raw
```

```python
np.zeros(X_raw.shape[1])
```

```python
y = y.reshape(-1, 1)
y
```

```python
def train(X, y, lr=0.0001, epochs=200):
    W = np.zeros((X.shape[1], 1))
    b = np.zeros((1, 1))
    losses = []
    for _ in range(epochs + 1):
        pred = X @ W + b
        err  = pred - y
        loss = (err**2).mean()
        losses.append(loss)
        W -= lr * (2/len(y)) * X.T @ err
        b -= lr * (2/len(y)) * err.mean()
        if _ % 50 == 0:
            print(f"Without norm — epoch {_}: {loss}")
    return losses
```

```python
X_no_norm = X_raw.res
losses_raw = train(X_no_norm, y, lr=0.000000001)
print(f"Without norm — epoch 0: {losses_raw[0]:.0f}, epoch 200: {losses_raw[-1]:.0f}")
```

```python
X_norm = (X_raw - X_raw.mean(axis=0))/ (X_raw.std(axis=0))
losses_raw = train(X_norm, y, lr = 0.18)
```
